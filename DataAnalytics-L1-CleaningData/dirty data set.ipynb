```python
# Clean column names
df.columns = (
    df.columns
      .str.replace('\xa0', ' ', regex=False)
      .str.strip()
)

print(df.columns.tolist())
```

```python
# Clean ranking values and handle missing values
for col in ['Peak', 'All Time Peak']:
    # Extract the actual rank from values such as "1[4]" or "2[7]"
    df[col] = pd.to_numeric(
        df[col].astype('string').str.extract(r'(\d+)')[0],
        errors='coerce'
    )

    # Fill missing rankings with the most common rank
    df[col] = df[col].fillna(df[col].mode().iloc[0])

print(df.isnull().sum())
```

## Missing Data Handling

Missing values were found only in the Peak and all time peak columns. Since these columns represent discrete ranking values, the missing values were filled using the mode rather than the mean. This preserves realistic ranking values without removing rows from the dataset.

```python
# Check and remove duplicate rows
duplicates_before = df.duplicated().sum()

df = df.drop_duplicates()

duplicates_after = df.duplicated().sum()

print("Duplicates before:", duplicates_before)
print("Duplicates after:", duplicates_after)
```

## Duplicate Removal

Duplicate rows were checked and removed to prevent repeated records from affecting the analysis. The number of duplicates before and after cleaning was documented.

```python
# Clean text columns
text_columns = ['Artist', 'Tour title', 'Year(s)', 'Ref.']

for col in text_columns:
    df[col] = df[col].astype('string').str.strip()

# Remove citation symbols/references from tour titles
df['Tour title'] = (
    df['Tour title']
    .str.replace(r'\[[^\]]+\]', '', regex=True)
    .str.replace('†', '', regex=False)
    .str.replace('‡', '', regex=False)
    .str.strip()
)

# Clean monetary columns
money_columns = [
    'Actual gross',
    'Adjusted gross (in 2022 dollars)',
    'Average gross'
]

for col in money_columns:
    df[col] = (
        df[col]
        .astype('string')
        .str.replace('$', '', regex=False)
        .str.replace(',', '', regex=False)
    )

    df[col] = pd.to_numeric(df[col], errors='coerce')

df.head()
```

## Data Standardisation

Text values were stripped of unnecessary spaces and tour titles were cleaned by removing reference symbols and citation markers. Monetary columns were standardised by removing dollar signs and commas so they could be converted into numerical values.

```python
print(df.dtypes)
print(df.isnull().sum())
```

```python
# Fill missing Actual gross values using Shows × Average gross
df['Actual gross'] = df['Actual gross'].fillna(
    df['Shows'] * df['Average gross']
)

print(df.isnull().sum())
```

### Additional Missing Value Handling

Two values in Actual gross became missing during numeric conversion because of inconsistent formatting. They were reconstructed using the number of shows multiplied by the average gross per show, preserving the relationship between the existing variables.

```python
numeric_columns = [
    'Actual gross',
    'Adjusted gross (in 2022 dollars)',
    'Shows',
    'Average gross'
]

for col in numeric_columns:
    Q1 = df[col].quantile(0.25)
    Q3 = df[col].quantile(0.75)
    IQR = Q3 - Q1

    lower_bound = Q1 - 1.5 * IQR
    upper_bound = Q3 + 1.5 * IQR

    outliers = df[(df[col] < lower_bound) | (df[col] > upper_bound)]

    print(f"{col}: {len(outliers)} outliers")
```

## Outlier Detection

The IQR method identified 2 outliers in Actual gross, 1 in Adjusted gross (in 2022 dollars), 1 in Shows, and 2 in Average gross.

These outliers were retained because extremely successful tours can legitimately generate much higher revenue or have more shows than typical tours. Removing them could eliminate meaningful real-world observations.

```python
# Correct data types

df['Rank'] = df['Rank'].astype('int64')
df['Peak'] = df['Peak'].astype('int64')
df['All Time Peak'] = df['All Time Peak'].astype('int64')

df['Actual gross'] = df['Actual gross'].astype(float)
df['Adjusted gross (in 2022 dollars)'] = df['Adjusted gross (in 2022 dollars)'].astype(float)
df['Average gross'] = df['Average gross'].astype(float)

df['Shows'] = df['Shows'].astype('int64')

df['Artist'] = df['Artist'].astype('string')
df['Tour title'] = df['Tour title'].astype('string')
df['Year(s)'] = df['Year(s)'].astype('string')
df['Ref.'] = df['Ref.'].astype('string')

print(df.dtypes)
```

## Data Type Correction

Ranking and show-count columns were converted to integer types, monetary columns were converted to float, and descriptive text columns were stored as strings. This ensures that each variable has an appropriate data type for analysis.

```python
# BEFORE vs AFTER summary

after_rows = len(df)
after_nulls = df.isnull().sum().sum()
after_duplicates = df.duplicated().sum()

summary = pd.DataFrame({
    'Metric': ['Row Count', 'Null Values', 'Duplicate Rows'],
    'Before': [before_rows, before_nulls, before_duplicates],
    'After': [after_rows, after_nulls, after_duplicates]
})

summary
```
